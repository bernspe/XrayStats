# Find optimal bspline reconstruction method

```python
import pandas as pd
import numpy as np
df4d=pd.read_pickle('data/df4d_corrected_spine_measurements.pkl')
#df=df4d[['id','uuid','m_uuid','nodes']]
def getNodes(xray_measurement_data):
    m=xray_measurement_data
    nodes = [(i['x'],i['y']) for i in m['stateObject']['nodes']]
    return nodes
df4d['nodes']=df4d['xray_measurement_data'].apply(getNodes)
df=df4d[['id','nodes']]
# limit dataset to those with groupby(id) count of 3
df=df.groupby('id').filter(lambda x: len(x) == 3)
```

```python
from spline_optimization.spline_optimization import SplineOptimizer
optimizer = SplineOptimizer(df, id_column='id', nodes_column='nodes', sort_axis=1)
# Optimiere
results = optimizer.optimize_all(verbose=False)

# Erstelle Bericht
report = optimizer.generate_report()
print(report)
```
